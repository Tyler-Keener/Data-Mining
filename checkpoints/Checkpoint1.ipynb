## (D) Exploratory Data Analysis (Selected Dataset Only)

Perform EDA on the selected dataset:
- Distribution of basket sizes
- Frequency of top items
- Sparsity of item co-occurrence
- Temporal gaps between transactions
- Initial observations motivating external techniques

````python
# Needed imports
import warnings
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# -------- Data Loading --------

# Load feature names
features = pd.read_csv('data/NUSW-NB15_features.csv', encoding='cp1252')
column_names = features['Name'].tolist()

# List of CSV files to load
csv_files = [
    'data/UNSW-NB15_1.csv',
    'data/UNSW-NB15_2.csv',
    'data/UNSW-NB15_3.csv',
    'data/UNSW-NB15_4.csv'
]

# Load with correct column names
data = pd.concat(
    (pd.read_csv(file, names=column_names, header=None) for file in csv_files),
    ignore_index=True
)

print(f"Data loaded successfully. Shape: {data.shape}")

print("\nData Information:")
print(data.info())

print("\nFirst 5 rows of the dataset:")
display(data.head())

# -------- Data Preprocessing --------

# Check for duplicates
duplicate_count = data.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}\n")

# Drop duplicates if any
if duplicate_count > 0:
    data = data.drop_duplicates()
    print(f"Duplicates dropped. New shape: {data.shape}\n")

# Check for missing values
missing_values = data.isnull().sum()
print("Missing values in each column:\n", missing_values[missing_values > 0])

# Fill missing values for specific columns
data['is_ftp_login'].fillna(0, inplace=True)
data['ct_flw_http_mthd'].fillna(0, inplace=True)
data['attack_cat'].fillna('Normal', inplace=True)

# Data type conversions
data['Stime'] = pd.to_numeric(data['Stime'], errors='coerce')
data['Stime'] = pd.to_datetime(data['Stime'], unit='s', errors='coerce')
data['Ltime'] = pd.to_numeric(data['Ltime'], errors='coerce')
data['Ltime'] = pd.to_datetime(data['Ltime'], unit='s', errors='coerce')

categorical_columns = ['proto','service', 'state', 'attack_cat']
for col in categorical_columns:
    data[col] = data[col].astype('category')

data['sport'] = pd.to_numeric(data['sport'], errors='coerce').astype('Int64')
data['dsport'] = pd.to_numeric(data['dsport'], errors='coerce').astype('Int64')
data = data.dropna(subset=['sport', 'dsport'])

data['ct_ftp_cmd'] = pd.to_numeric(data['ct_ftp_cmd'], errors='coerce').fillna(0).astype(int)
data['is_ftp_login'] = pd.to_numeric(data['is_ftp_login'], errors='coerce').fillna(0).astype(int)
data['ct_flw_http_mthd'] = pd.to_numeric(data['ct_flw_http_mthd'], errors='coerce').fillna(0).astype(int)

# Strip whitespace/convert to lowercase the categorical columns
for col in categorical_columns:
    data[col] = data[col].str.strip().str.lower()

# -------- Exploratory Data Analysis --------

# Class distribution: 'label' (0=Normal, 1=Attack)
plt.figure(figsize=(6,4))
data['Label'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Class Distribution: Normal vs. Attack')
plt.xlabel('Class')
plt.ylabel('Count')
plt.xticks([0,1], ['Normal', 'Attack'], rotation=0)
plt.show()
print(data['Label'].value_counts(normalize=True))

# Attack category distribution
official_categories = [
    'fuzzers', 'analysis', 'backdoors', 'dos', 'exploits',
    'generic', 'reconnaissance', 'shellcode', 'worms'
]
attack_data = data[data['attack_cat'].isin(official_categories)]
attack_counts = attack_data['attack_cat'].value_counts().reindex(official_categories, fill_value=0)

plt.figure(figsize=(10,5))
attack_counts.plot(kind='bar', color='teal')
plt.title('Attack Category Distribution')
plt.xlabel('Attack Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
print(attack_counts / attack_counts.sum())

# Co-occurrence by source IP
co_occurrence = pd.crosstab(attack_data['srcip'], attack_data['attack_cat'])
co_occurrence = co_occurrence[co_occurrence.sum(axis=1) > 1]
sns.heatmap(co_occurrence.T, cmap='Blues', cbar=True)
plt.title('Attack Category Co-occurrence by Source IP')
plt.xlabel('Source IP')
plt.ylabel('Attack Category')
plt.show()
````
