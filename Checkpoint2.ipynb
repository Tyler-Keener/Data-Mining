### Research Question 1

- **Motivation:**  
  Exploratory data analysis shows that the dataset has significant class imbalance, with some attack categories appearing very rarely compared to normal traffic. Anomaly detection methods are well suited for this scenario because they are designed to identify rare or unusual patterns without requiring balanced labeled data.

- **Non-triviality:**  
  Traditional supervised classification methods may struggle when attack classes are extremely underrepresented. Evaluating unsupervised anomaly detection techniques provides insight into whether meaningful attack detection is possible without relying heavily on labeled examples.

- **Feasibility:**  
  Algorithms such as Isolation Forest, Local Outlier Factor, and One-Class SVM are widely available in libraries like `scikit-learn` and are efficient enough to run on large tabular datasets. The dataset is already structured as feature vectors, making it straightforward to apply these methods.

- **Risks:**  
  - High false positive rates may occur because normal traffic can vary significantly.  
  - Some algorithms may be sensitive to hyperparameter choices.  
  - Performance evaluation can be challenging when attacks are extremely rare.

---

### Research Question 2

- **Motivation:**  
  Clustering can reveal hidden structure in network traffic by grouping similar connections together. If attack types form distinct clusters in feature space, clustering methods could help identify unknown or emerging attack patterns without relying on labeled data.

- **Non-triviality:**  
  Network traffic often contains overlapping feature distributions between normal and malicious behavior. This makes it unclear whether clustering algorithms will naturally separate attack types or whether additional feature engineering is required.

- **Feasibility:**  
  Clustering algorithms such as K-Means, DBSCAN, and hierarchical clustering are well-supported in standard data mining libraries and can be applied directly to the numerical feature space after preprocessing.

- **Risks:**  
  - High-dimensional feature spaces may reduce clustering effectiveness.  
  - Some clustering algorithms require careful parameter tuning.  
  - Clusters may not correspond directly to attack labels.

---

### Research Question 3

- **Motivation:**  
  Machine learning–based intrusion detection systems are increasingly deployed in cybersecurity environments. However, recent research shows that deep learning models can be vulnerable to adversarial perturbations that slightly modify input features to evade detection.

- **Non-triviality:**  
  Traditional data mining techniques typically focus on predictive performance but do not evaluate model robustness against adversarial manipulation. Studying adversarial attacks introduces an additional layer of security analysis beyond standard classification accuracy.

- **Feasibility:**  
  Deep neural networks for tabular data can be implemented using frameworks such as PyTorch or TensorFlow. Adversarial attack methods like FGSM and PGD are well-documented and can be implemented using existing adversarial machine learning libraries.

- **Risks:**  
  - Adversarial attacks must be carefully constrained to ensure feature perturbations remain realistic for network traffic data.  
  - Implementing adversarial attacks may increase computational complexity.  
  - Some features (e.g., categorical variables) are harder to perturb meaningfully.
